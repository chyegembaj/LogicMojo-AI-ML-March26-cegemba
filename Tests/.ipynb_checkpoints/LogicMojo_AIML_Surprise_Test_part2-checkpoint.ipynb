Problem 2: Multiple Linear Regression - House Price Prediction
A real estate dataset contains details of five houses along with their sale price.
Task: Train a multiple linear regression model using Area and Bedrooms as independent variables to predict Price. Report the model coefficients and intercept, and predict the price of a house with Area = 2000 sq ft and 3 Bedrooms.



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
#Price in Lakhs
data = pd.DataFrame({
    "SquareFeets":[1000, 1500, 1800, 2200, 2500],
    "Bedrooms": [2, 3, 3, 4, 4],
    "Price(in Lakhs)": [50, 65, 72, 85, 95]
    
})
print(data)
X = data[['SquareFeets', 'Bedrooms']]
y = data['Price(in Lakhs)']

model = LinearRegression()
model.fit(X, y)
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
prediction = model.predict([[2000, 3]])
print(f"Predicted Price:, ${prediction[0]:,.2f}")
```

Problem 3: Polynomial Regression
The following data points are collected from an experiment where the output y depends on input x.
Task: Fit a polynomial regression model of degree 2 to the given data. Report the fitted model equation and predict the value of y when x = 6.



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 8, 18, 32, 50])

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

x_new = np.array([[6]])
x_new_poly = poly.transform(x_new)
prediction = model.predict(x_new_poly)

print("Slope:", model.coef_)
print("Intercept:", model.intercept_)
print("Predicted: ", prediction[0] )

```

Problem 5: Descriptive Statistics
The following dataset represents the ages of 10 participants in a workshop.
Ages: 12, 15, 14, 10, 18, 20, 14, 16, 15, 22

Task: Compute the mean, median, mode, variance, and standard deviation of the dataset.


```python
import numpy as np
import statistics as stats

ages = np.array([12, 15, 14, 10, 18, 20, 14, 16, 15, 22])

mean = np.mean(ages)
median = np.median(ages)
mode = stats.multimode(ages)
variance = np.var(ages)
std = np.std(ages)

print("mean:", mean )
print("median:", median )
print("mode:", mode)
print("variance:", variance )
print("Standard Deviation:", std )

```

Problem 6: Z-Score Calculation
A dataset of exam scores is given below.
Scores 45, 50, 55, 60, 65, 70, 75

Task: Compute the mean and standard deviation of the dataset. Then calculate the z-score for the value 75, and state whether it should be considered an outlier if any value with |z| > 2 is treated as an outlier.

```python
import numpy as np
import statistics as stats

scores = np.array([45, 50, 55, 60, 65, 70, 75])
mean = np.mean(scores)
std = np.std(scores)

print("mean:", mean )
print("Standard Deviation:", std )

x = 75
z_score = (x - mean)/std
print("z-score: ", z_score)

#|z_score| is < 2, 75 is not considered an outlier 
```

Problem 7: Correlation Coefficient
A company recorded its monthly advertising spend and the corresponding sales.
Advertising Spend (in Rs '000) 200, 250, 300, 350, 400
Sales (units) 20, 24, 27, 33, 38

Task: Compute the Pearson correlation coefficient between Advertising Spend and Sales. Interpret the strength and direction of the relationship between the two variables.

```python
import numpy as np
import statistics as stats
import pandas as pd


advertising = np.array([200, 250, 300, 350, 400])
sales = np.array([20, 24, 27, 33, 38])

correlation_coef = np.corrcoef(advertising, sales)[0, 1]

print("Correlation coefficient: ", correlation_coef)
    
#Since 0.993 Correlation coefficient is close to 1, there is a strong postive linear
#relationship. As advertising spend increases, sales also increases
```

Problem 8: Binomial Probability
In a quality control process at a manufacturing unit, the probability that a randomly selected item is defective is 0.1. A random sample of 8 items is inspected.
Task: Using the Binomial distribution, calculate the probability that exactly 2 items in the sample of 8 are defective.


```python
from scipy.stats import binom

p = 0.1
n = 8
x = 2

probability = binom.pmf(x, n, p)
print("probability :", probability)
```
